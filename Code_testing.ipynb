```python
import numpy as np
```

```python
sample_size = 100_000
batch_size = 256

starting_indices = np.random.choice(int(1e8), sample_size//batch_size, replace=False)
indices = []
for starting_index in starting_indices:
    indices.append(np.arange(starting_index, starting_index + batch_size))
indices = np.concatenate(indices)
indices.sort()
```
